# **Sistemas de ecuaciones**

se escribe de la siguiente forma: los sistemas de las ecuaciones lineales



Un sistema de n ecuaciones con n incógnitas se puede escribir de la siguiente forma:



Lo que se puede reescribir de una forma compactada utilizando **Ax = b** donde **A** es una matriz de dimension *n x n* y *x* y *b* son dos vectores columna de longitud *n*. En esta ecuacion 



```python
import numpy as np

A = np.matrix([[2, 3],[1, -2]])
b = np.matrix([[8],[-10]])

x = (A**-1)*b
```

```python
if np.linalg.det(A) == 0:
    x = None
    print("No se puede resolver")
else:
    x = (A**-1)*b
print("Esto es A: \n ", A)
print()
print("Esto es b: \n ",b)
```

```python
print("la inversa es: \n",x)
```
